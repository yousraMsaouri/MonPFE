### charger les trois fichiers et les fusionner en un seul gros DataFrame

```python
import pandas as pd
import os

# 1. Définition du chemin vers le dossier data
DATA_DIR = os.path.join("..", "data")

fichiers = ["TotalFrance.xlsx", "TotalMaroc.xlsx", "TotalSpain.xlsx"]
all_dfs = []

print("--- NOUVELLE TENTATIVE DE FUSION (COLONNES CORRIGÉES) ---")

# 2. Boucle pour lire et harmoniser les colonnes
for nom_fichier in fichiers:
    chemin_complet = os.path.join(DATA_DIR, nom_fichier)
    
    if not os.path.exists(chemin_complet):
        print(f"⚠️ Fichier introuvable : {nom_fichier}")
        continue
        
    print(f"Lecture et nettoyage des en-têtes de : {nom_fichier}...")
    df_temp = pd.read_excel(chemin_complet)
    
    # Nettoyage automatique des espaces au début/fin des noms de colonnes pour ce fichier
    df_temp.columns = df_temp.columns.str.strip()
    
    all_dfs.append(df_temp)

# 3. Fusion verticale des 3 pays
datainternational = pd.concat(all_dfs, ignore_index=True)

print("\n--- FUSION REUSSIE ! ---")
print(f"Nouvelles dimensions : {datainternational.shape[0]} lignes x {datainternational.shape[1]} colonnes")
print("\nListe finale des colonnes conservées (il ne doit plus y avoir de doublons) :")
print(list(datainternational.columns))

# Aperçu
datainternational.head()
```

### Agregation des Promos 

```python
import numpy as np
import pandas as pd
import os

# Configuration du dossier de sauvegarde
DATA_DIR = os.path.join("..", "data")

print(f"Nombre de lignes initiales : {datainternational.shape[0]}")

# 1. Nettoyage strict des codes produits parasites -1
datainternational = datainternational[datainternational['Material Code'] != -1].copy()
datainternational = datainternational[datainternational['Material Code'] != '-1'].copy()

# 2. Fonction de classification stricte pour 'Order Reason'
def classifier_motif_strict(valeur):
    texte = str(valeur).strip()
    
    # Cas vides (NaN ou texte 'nan') -> Commande standard
    if pd.isna(valeur) or texte == '' or texte.lower() == 'nan':
        return 'Standard order'
    elif texte == 'Standard order':
        return 'Standard order'
    elif texte == 'Promotion order':
        return 'Promotion order'
    # Tout le reste (Quotation, Showroom, etc.) -> Autre
    else:
        return 'Autre'

# Application de la classification
datainternational['Order_Reason_Clean'] = datainternational['Order Reason'].apply(classifier_motif_strict)

# 3. Création du flag binaire Is_Promo (1 = Promo, 0 = Non Promo)
datainternational['Is_Promo'] = datainternational['Order_Reason_Clean'].apply(lambda x: 1 if x == 'Promotion order' else 0)

print(f"Nombre de lignes après nettoyage : {datainternational.shape[0]}")

# --- ÉTAPE DE SAUVEGARDE DU FICHIER DÉTAILLÉ (Avant Agrégation) ---
OUTPUT_DETAILLE = os.path.join(DATA_DIR, "datainternational_detaille_clean.csv")
print("\nSauvegarde du fichier détaillé en cours...")
datainternational.to_csv(OUTPUT_DETAILLE, index=False, encoding='utf-8')
print(f"✅ Fichier détaillé sauvegardé avec succès sous : {OUTPUT_DETAILLE}")

print("\n--- VÉRIFICATION DES CATÉGORIES ---")
print(datainternational['Order_Reason_Clean'].value_counts())

# Aperçu
datainternational[['Material Code', 'Order Reason', 'Order_Reason_Clean', 'Is_Promo', 'Sales Quantity']].head(5)
```

### Agregation Final des meme ventes 

```python
import pandas as pd
import os

# 1. Configuration du dossier data
DATA_DIR = os.path.join("..", "data")

print(f"Nombre de lignes avant agrégation finale : {datainternational.shape[0]}")

# 2. Définition de toutes les colonnes clés (On inclut Is_Promo pour séparer les promos des non-promos)
colonnes_regroupement = [
    'Country', 'Year', 'Month Name', 'Channel', 
    'Material Code', 'Material Code Name', 
    'Category', 'Sous Category', 'Collection',
    'Is_Promo'  # <-- Crucial : sépare l'agrégation de Is_Promo = 0 et Is_Promo = 1
]

# 3. Agrégation mensuelle : On calcule la somme NETTE des Sales Quantity
datainternational_final = datainternational.groupby(colonnes_regroupement, as_index=False).agg({
    'Sales Quantity': 'sum'
})

print(f"Nombre de lignes après agrégation finale : {datainternational_final.shape[0]}")

# 4. Tri chronologique et logique pour que le dataset soit parfait
datainternational_final = datainternational_final.sort_values(
    by=['Country', 'Material Code', 'Channel', 'Year', 'Month Name', 'Is_Promo']
).reset_index(drop=True)

# 5. Sauvegarde physique du dataset final épuré
OUTPUT_FINAL_PATH = os.path.join(DATA_DIR, "datainternational_final.csv")
datainternational_final.to_csv(OUTPUT_FINAL_PATH, index=False, encoding='utf-8')

print(f"\n🎉 Félicitations ! Votre fichier final de PFE est agrégé et sauvegardé sous : datainternational_final.csv")

# Aperçu du résultat final
datainternational_final.head(10)
```

## Créer une vraie colonne Date standardisée

```python
import pandas as pd
import os

# 1. Configuration des chemins et rechargement du fichier agrégé
DATA_DIR = os.path.join("..", "data")
PATH_AGREGE = os.path.join(DATA_DIR, "datainternational_final.csv")

df_pfe = pd.read_csv(PATH_AGREGE)

print(f"Structure initiale : {df_pfe.shape[0]} lignes x {df_pfe.shape[1]} colonnes")

# 2. Dictionnaire de correspondance pour convertir les mois textuels en numéros
# (Gère à la fois l'anglais et le français au cas où)
mois_mapping = {
    'January': '01', 'February': '02', 'March': '03', 'April': '04',
    'May': '05', 'June': '06', 'July': '07', 'August': '08',
    'September': '09', 'October': '10', 'November': '11', 'December': '12',
    'Janvier': '01', 'Février': '02', 'Mars': '03', 'Avril': '04',
    'Mai': '05', 'Juin': '06', 'Juillet': '07', 'Août': '08',
    'Septembre': '09', 'Octobre': '10', 'Novembre': '11', 'Décembre': '12'
}

# 3. Création de la colonne de date standardisée (Format : YYYY-MM-01)
# On applique le mapping sur 'Month Name'
df_pfe['Month_Num'] = df_pfe['Month Name'].map(mois_mapping)

# On combine l'année et le numéro du mois pour créer la chaîne de caractères (ex: "2022-03-01")
df_pfe['Date_String'] = df_pfe['Year'].astype(str) + '-' + df_pfe['Month_Num'] + '-01'

# On transforme cette chaîne en un vrai type Date (datetime) reconnu par Python
df_pfe['Date'] = pd.to_datetime(df_pfe['Date_String'])

# 4. Suppression des colonnes intermédiaires devenues inutiles pour garder le fichier propre
df_pfe = df_pfe.drop(columns=['Month_Num', 'Date_String'])

# 5. Tri chronologique et logique parfait du dataset complet
df_pfe = df_pfe.sort_values(
    by=['Country', 'Material Code', 'Channel', 'Date', 'Is_Promo']
).reset_index(drop=True)

# 6. Sauvegarde du fichier mis à jour avec la variable Date
df_pfe.to_csv(PATH_AGREGE, index=False, encoding='utf-8')

print("✅ Colonne 'Date' créée avec succès au format standardisé ! Fichier mis à jour.")
print(f"Structure actuelle : {df_pfe.shape[0]} lignes x {df_pfe.shape[1]} colonnes")

# Aperçu pour valider le résultat visuellement
df_pfe[['Country', 'Year', 'Month Name', 'Date', 'Material Code', 'Sales Quantity']].head(10)
```

## Creation des lags 

```python
import pandas as pd
import os

# 1. Configuration des chemins et rechargement du fichier final agrégé
DATA_DIR = os.path.join("..", "data")
PATH_AGREGE = os.path.join(DATA_DIR, "datainternational_final.csv")

df_pfe = pd.read_csv(PATH_AGREGE)

# Tri parfait par groupe chronologique avant d'effectuer les décalages (shifts)
df_pfe = df_pfe.sort_values(by=['Country', 'Material Code', 'Channel', 'Is_Promo', 'Date']).reset_index(drop=True)

print("Création du set complet de variables historiques (Lags)...")

# Groupe de caractéristiques uniques pour ne pas mélanger les pinceaux entre pays/SKU/canaux
group_features = ['Country', 'Material Code', 'Channel', 'Is_Promo']

# 2. Création des Lags à court terme
df_pfe['Sales_Qty_Lag_1'] = df_pfe.groupby(group_features)['Sales Quantity'].shift(1)
df_pfe['Sales_Qty_Lag_2'] = df_pfe.groupby(group_features)['Sales Quantity'].shift(2)
df_pfe['Sales_Qty_Lag_3'] = df_pfe.groupby(group_features)['Sales Quantity'].shift(3)

# 3. Création des Lags à moyen et long terme (Saisonnalité)
df_pfe['Sales_Qty_Lag_6'] = df_pfe.groupby(group_features)['Sales Quantity'].shift(6)
df_pfe['Sales_Qty_Lag_12'] = df_pfe.groupby(group_features)['Sales Quantity'].shift(12)

# 4. Sauvegarde physique du fichier enrichi
df_pfe.to_csv(PATH_AGREGE, index=False, encoding='utf-8')

print("✅ Toutes les variables (Lag 1, 2, 3, 6 et 12) ont été ajoutées et sauvegardées avec succès !")

# Aperçu pour s'assurer que les nouvelles colonnes sont bien présentes tout à droite
df_pfe[['Country', 'Date', 'Material Code', 'Sales Quantity', 'Sales_Qty_Lag_1', 'Sales_Qty_Lag_6', 'Sales_Qty_Lag_12']].head(5)
```

```python
import pandas as pd
import os

DATA_DIR = os.path.join("..", "data")
PATH_AGREGE = os.path.join(DATA_DIR, "datainternational_final.csv")
df_pfe = pd.read_csv(PATH_AGREGE)
df_pfe['Date'] = pd.to_datetime(df_pfe['Date'])

print(f"Nombre de lignes avant reconstruction du calendrier : {df_pfe.shape[0]}")

# 1. Suppression des anciennes colonnes Lags erronées pour repartir sur une base propre
lags_cols = ['Sales_Qty_Lag_1', 'Sales_Qty_Lag_2', 'Sales_Qty_Lag_3', 'Sales_Qty_Lag_6', 'Sales_Qty_Lag_12']
df_pfe = df_pfe.drop(columns=[c for c in lags_cols if c in df_pfe.columns])

# 2. Création d'un calendrier complet pour chaque groupe unique
group_features = ['Country', 'Material Code', 'Material Code Name', 'Channel', 'Category', 'Sous Category', 'Collection', 'Is_Promo']

df_list = []
for name, group in df_pfe.groupby(group_features, as_index=False):
    # On génère tous les mois sans exception entre la date de début et de fin de ce produit
    all_months = pd.date_range(start=group['Date'].min(), end=group['Date'].max(), freq='MS')
    
    # On reconstruit la structure temporelle
    group = group.set_index('Date').reindex(all_months)
    group.index.name = 'Date'
    
    # On remplit les informations fixes pour les mois ajoutés
    for i, col in enumerate(group_features):
        group[col] = name[i]
        
    # S'il n'y a pas eu de vente ce mois-ci, la quantité réelle vaut 0
    group['Sales Quantity'] = group['Sales Quantity'].fillna(0)
    df_list.append(group.reset_index())

df_reconstructed = pd.concat(df_list, ignore_index=True)
print(f"Nombre de lignes après reconstruction (les mois vides sont à 0) : {df_reconstructed.shape[0]}")

# 3. Recalcul des Lags parfaits sur le calendrier continu
df_reconstructed = df_reconstructed.sort_values(by=group_features + ['Date']).reset_index(drop=True)

df_reconstructed['Sales_Qty_Lag_1'] = df_reconstructed.groupby(group_features)['Sales Quantity'].shift(1)
df_reconstructed['Sales_Qty_Lag_2'] = df_reconstructed.groupby(group_features)['Sales Quantity'].shift(2)
df_reconstructed['Sales_Qty_Lag_3'] = df_reconstructed.groupby(group_features)['Sales Quantity'].shift(3)
df_reconstructed['Sales_Qty_Lag_6'] = df_reconstructed.groupby(group_features)['Sales Quantity'].shift(6)
df_reconstructed['Sales_Qty_Lag_12'] = df_reconstructed.groupby(group_features)['Sales Quantity'].shift(12)

# 4. Sauvegarde finale
df_reconstructed.to_csv(PATH_AGREGE, index=False, encoding='utf-8')
print("✅ Calendrier reconstruit et Lags temporels exacts enregistrés !")

# Aperçu
df_reconstructed[['Country', 'Date', 'Material Code', 'Sales Quantity', 'Sales_Qty_Lag_1', 'Sales_Qty_Lag_2']].head(10)
```

### ajout de saisons 

```python
import pandas as pd
import os

# 1. Configuration des chemins et rechargement du fichier
DATA_DIR = os.path.join("..", "data")
PATH_AGREGE = os.path.join(DATA_DIR, "datainternational_final.csv")

df_pfe = pd.read_csv(PATH_AGREGE)
df_pfe['Date'] = pd.to_datetime(df_pfe['Date'])

print("Extraction des saisons...")

# 2. Fonction de mapping pour associer chaque numéro de mois à sa saison
def extraire_saison(mois):
    if mois in [12, 1, 2]:
        return 'Hiver'
    elif mois in [3, 4, 5]:
        return 'Printemps'
    elif mois in [6, 7, 8]:
        return 'Ete'
    elif mois in [9, 10, 11]:
        return 'Automne'

# 3. Application de la fonction en utilisant la colonne 'Month_Num' créée à l'étape précédente
df_pfe['Season'] = df_pfe['Date'].dt.month.apply(extraire_saison)

# 4. Sauvegarde physique du fichier enrichi
df_pfe.to_csv(PATH_AGREGE, index=False, encoding='utf-8')

print("✅ Colonne 'Season' (Saison) ajoutée avec succès !")

# Aperçu pour vérifier la cohérence temporelle
df_pfe[['Country', 'Date', 'Month Name', 'Quarter', 'Season', 'Sales Quantity']].head(12)
```

### Moyenne et Somme Glissantes

```python
import pandas as pd
import os

# 1. Configuration des chemins et rechargement du fichier
DATA_DIR = os.path.join("..", "data")
PATH_AGREGE = os.path.join(DATA_DIR, "datainternational_final.csv")

df_pfe = pd.read_csv(PATH_AGREGE)
df_pfe['Date'] = pd.to_datetime(df_pfe['Date'])

# Tri chronologique parfait par groupe pour s'assurer que le glissement se fait dans le bon ordre
group_features = ['Country', 'Material Code', 'Channel', 'Is_Promo']
df_pfe = df_pfe.sort_values(by=group_features + ['Date']).reset_index(drop=True)

print("Calcul des variables glissantes (Rolling Features) sur 3 mois...")

# 2. MOYENNE GLISSANTE 3 MOIS (en excluant le mois en cours via shift(1))
df_pfe['Sales_Qty_Rolling_Mean_3M'] = (
    df_pfe.groupby(group_features)['Sales Quantity']
    .transform(lambda x: x.shift(1).rolling(window=3, min_periods=1).mean())
)

# 3. SOMME GLISSANTE 3 MOIS (en excluant le mois en cours via shift(1))
df_pfe['Sales_Qty_Rolling_Sum_3M'] = (
    df_pfe.groupby(group_features)['Sales Quantity']
    .transform(lambda x: x.shift(1).rolling(window=3, min_periods=1).sum())
)

# 4. Sauvegarde physique du fichier enrichi
df_pfe.to_csv(PATH_AGREGE, index=False, encoding='utf-8')

print("✅ Colonnes 'Sales_Qty_Rolling_Mean_3M' et 'Sales_Qty_Rolling_Sum_3M' ajoutées avec succès !")

# Aperçu du résultat pour observer le calcul glissant
df_pfe[['Country', 'Date', 'Material Code', 'Sales Quantity', 'Sales_Qty_Rolling_Mean_3M', 'Sales_Qty_Rolling_Sum_3M']].head(10)
```

## intensite de crises 

```python
import pandas as pd
import os

# 1. Configuration des chemins et rechargement
DATA_DIR = os.path.join("..", "data")
PATH_AGREGE = os.path.join(DATA_DIR, "datainternational_final.csv")
df_pfe = pd.read_csv(PATH_AGREGE)
df_pfe['Date'] = pd.to_datetime(df_pfe['Date'])

print("Correction de l'intensité des crises (Strictement 0, 0.5, 1)...")

# 2. Nouvelle fonction respectant strictement les 3 paliers demandés
def calculer_intensite_crise_stricte(row):
    date = row['Date']
    pays = str(row['Country']).lower()
    
    # Année 2022 (Vagues de crises initiales)
    if date.year == 2022:
        if pays in ['china', 'chine']:
            return 1.0  # Confinements maximaux
        return 0.5     # Intensité moyenne pour la France / Maroc / Espagne
        
    # Année 2023 (Pic de la crise énergétique européenne)
    elif date.year == 2023:
        if pays in ['france', 'spain', 'espagne']:
            return 1.0  # Intensité maximale sur l'industrie en Europe
        return 0.5     # Intensité moyenne pour le Maroc / Chine (impacts indirects)
        
    # Avant 2022 et À partir de 2024 : Retour au calme / Pas de crise majeure enregistrée
    return 0.0

# 3. Application du mapping strict
df_pfe['Crisis_Intensity'] = df_pfe.apply(calculer_intensite_crise_stricte, axis=1)

# 4. Sauvegarde
df_pfe.to_csv(PATH_AGREGE, index=False, encoding='utf-8')

print("✅ Colonne 'Crisis_Intensity' corrigée avec succès !")
# Vérification de contrôle de la distribution
print(df_pfe.groupby(['Country', df_pfe['Date'].dt.year])['Crisis_Intensity'].mean())
```

```python
import pandas as pd
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Rechargement du fichier
DATA_DIR = os.path.join("..", "data")
PATH_AGREGE = os.path.join(DATA_DIR, "datainternational_final.csv")
df_pfe = pd.read_csv(PATH_AGREGE)

# Filtrage sur tes 3 pays cibles et uniformisation
pays_cibles = ['France', 'Morocco', 'Spain', 'Maroc', 'Espagne']
df_3_pays = df_pfe[df_pfe['Country'].str.title().isin(pays_cibles)].copy()
df_3_pays['Country'] = df_3_pays['Country'].replace({'Maroc': 'Morocco', 'Espagne': 'Spain'})

# NETTOYAGE : On ne garde que les ventes strictement positives pour le calcul du LOG
df_clean_dist = df_3_pays[df_3_pays['Sales Quantity'] >= 0].copy()

# 2. Calcul du LOG de manière sécurisée
df_clean_dist['Log_Sales_Quantity'] = np.log1p(df_clean_dist['Sales Quantity'])

# 3. Création d'une grille de graphiques séparés (3 lignes pour les 3 pays, 2 colonnes pour Brut vs Log)
liste_pays = ['France', 'Morocco', 'Spain']
fig, axes = plt.subplots(3, 2, figsize=(16, 15))

colors_dict = {'France': '#4c72b0', 'Morocco': '#dd8452', 'Spain': '#55a868'}

for i, pays in enumerate(liste_pays):
    df_sub = df_clean_dist[df_clean_dist['Country'] == pays]
    
    # --- Colonne 1 : Distribution Brute ---
    sns.histplot(data=df_sub, x='Sales Quantity', color=colors_dict[pays], kde=True, ax=axes[i, 0], bins=40)
    axes[i, 0].set_title(f"{pays} - Distribution Brute (Asymétrique)", fontsize=11, weight='bold')
    axes[i, 0].set_xlabel("Sales Quantity (Unités)")
    axes[i, 0].set_ylabel("Fréquence")
    
    # --- Colonne 2 : Distribution Log ---
    sns.histplot(data=df_sub, x='Log_Sales_Quantity', color=colors_dict[pays], kde=True, ax=axes[i, 1], bins=40)
    axes[i, 1].set_title(f"{pays} - Distribution Logarithmique (Symétrique)", fontsize=11, weight='bold')
    axes[i, 1].set_xlabel("Log(Sales Quantity + 1)")
    axes[i, 1].set_ylabel("Fréquence")

plt.suptitle("Analyse Comparative Individualisée des Distributions par Pays", fontsize=14, y=1.01, weight='bold')
plt.tight_layout()
plt.show()

# 4. Affichage des scores de Skewness corrigés sans NaN
print("📊 NOUVEAUX SCORES DE SKEWNESS (Nettoyés des valeurs négatives) :")
print("-" * 70)
for pays in liste_pays:
    df_sub = df_clean_dist[df_clean_dist['Country'] == pays]
    skew_brut = df_sub['Sales Quantity'].skew()
    skew_log = df_sub['Log_Sales_Quantity'].skew()
    print(f"🔹 {pays.upper():<8} -> Skewness Brut: {skew_brut:.2f} | Skewness Log: {skew_log:.2f}")
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import pandas as pd

# 1. Sélectionner uniquement les colonnes de ventes passées et moyennes mobiles de 'df_encoded'
colonnes_a_compresser = [
    'Sales_Qty_Lag_1', 'Sales_Qty_Lag_2', 'Sales_Qty_Lag_3', 
    'Sales_Qty_Lag_12', 'Sales_Qty_Lag_24', 
    'Sales_Qty_Rolling_Mean_3M', 'Sales_Qty_Rolling_Sum_3M'
]

X_correrele = df_encoded[colonnes_a_compresser]

# 2. Centrer et Réduire (Normalisation obligatoire pour la PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_correrele)

# 3. Application de la PCA pour créer 2 Composantes Principales (PC1 et PC2)
pca = PCA(n_components=2)
composantes_principales = pca.fit_transform(X_scaled)

# 4. Création du tableau avec les nouvelles colonnes magiques
df_pca = pd.DataFrame(data=composantes_principales, columns=['PC1', 'PC2'])

# 5. Calcul et affichage du pourcentage d'information conservée
variance_conservee = pca.explained_variance_ratio_.sum() * 100
print(f"✨ La PCA a compressé tes 7 variables en 2 colonnes avec succès !")
print(f"📈 Information conservée : {variance_conservee:.2f}% de la variance totale.")
```

```python
# 1. Fusionner les colonnes PC1 et PC2 dans ton tableau principal
df_final_pca = pd.concat([df_pfe.reset_index(drop=True), df_pca], axis=1)

# 2. Supprimer les 7 anciennes colonnes corrélées pour alléger le modèle
df_final_pca = df_final_pca.drop(columns=colonnes_a_compresser)

print("🚀 Étape réussie !")
print(f"Ancien nombre de colonnes : {df_pfe.shape[1]}")
print(f"Nouveau nombre de colonnes après PCA : {df_final_pca.shape[1]}")
print("\nVoici un aperçu de tes données prêtes pour l'entraînement :")
df_final_pca[['Country', 'Channel', 'Material Code', 'Sales Quantity', 'PC1', 'PC2']].head()
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from category_encoders import TargetEncoder
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("🚀 ÉTAPE 1 : Préparation des datasets (Intégration de la PCA)...")
# On supprime juste les lignes vides au cas où, mais tes données sont déjà propres
df_train_clean = df_train.dropna(subset=['Sales Quantity']).copy()
df_test_clean = df_test.dropna(subset=['Sales Quantity']).copy()

print(f"✅ Validation des données : Train = {df_train_clean.shape[0]} | Test = {df_test_clean.shape[0]}")

print("\n🚀 ÉTAPE 2 : Encodage robuste avec PC1 et PC2...")
# 📢 ATTENTION : Remplacement des anciens Lags par PC1 et PC2
features_model = [
    'Material Code', 'Country', 'Channel', 'Category', 'Sous Category', 'Collection', 'Season',
    'Is_Promo', 'Quarter', 'Month_Num', 'Is_Start_Year', 'Is_End_Year',
    'PC1', 'PC2', 'Crisis_Intensity'
]

y_train_log = np.log1p(df_train_clean['Sales Quantity'].clip(lower=0))
y_test_real = df_test_clean['Sales Quantity'].values

X_train_raw = df_train_clean[features_model].copy()
X_test_raw = df_test_clean[features_model].copy()

# 1. Target Encoding pour les fortes cardinalités
colonnes_target = ['Material Code', 'Sous Category', 'Collection']
te = TargetEncoder(cols=colonnes_target, smoothing=10.0)
X_train_encoded = te.fit_transform(X_train_raw, y_train_log)
X_test_encoded = te.transform(X_test_raw)

# 2. Label Encoding sécurisé contre les valeurs inconnues ('UNKNOWN')
for col in ['Country', 'Channel', 'Category', 'Season']:
    categories = {val: i for i, val in enumerate(X_train_raw[col].astype(str).unique())}
    X_train_encoded[col] = X_train_raw[col].astype(str).map(categories)
    X_test_encoded[col] = X_test_raw[col].astype(str).map(categories).fillna(-1).astype(int)

print("✅ Encodages sécurisés appliqués avec succès.")

print("\n🚀 ÉTAPE 3 : Normalisation globale (MinMax Scaling 0-1)...")
scaler = MinMaxScaler()
X_train_final = pd.DataFrame(scaler.fit_transform(X_train_encoded), columns=X_train_encoded.columns)
X_test_final = pd.DataFrame(scaler.transform(X_test_encoded), columns=X_test_encoded.columns)

print("✅ Normalisation validée.")

print("\n📐 ÉTAPE 4 : Redimensionnement 3D pour le LSTM...")
X_train_3D = np.reshape(X_train_final.values, (X_train_final.shape[0], 1, X_train_final.shape[1]))
X_test_3D = np.reshape(X_test_final.values, (X_test_final.shape[0], 1, X_test_final.shape[1]))

print(f"Shape Train 3D : {X_train_3D.shape} (Le nombre de features est passé à {X_train_3D.shape[2]})")

print("\n🧠 ÉTAPE 5 : Lancement du LSTM OPTIMISÉ (Learning Rate = 0.001)...")
model_lstm = Sequential([
    Input(shape=(X_train_3D.shape[1], X_train_3D.shape[2])), # S'adapte automatiquement aux dimensions de la PCA
    LSTM(128, return_sequences=True),
    Dropout(0.3),
    LSTM(64),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dense(1)
])

model_lstm.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), loss='mse')
early_stop = EarlyStopping(monitor='val_loss', patience=6, restore_best_weights=True)

# Entraînement
history = model_lstm.fit(
    X_train_3D, y_train_log,
    epochs=35,
    batch_size=256,
    validation_split=0.15,
    callbacks=[early_stop],
    verbose=1
)

print("\n🎯 ÉTAPE 6 : Évaluation finale du modèle...")
y_pred_log = model_lstm.predict(X_test_3D).flatten()
y_pred_real = np.clip(np.expm1(y_pred_log), 0, None)

print("\n🏆 VERDICT DU MODÈLE DEEP LEARNING (LSTM AVEC PCA) :")
print("-" * 60)
print(f"📈 R² LSTM   : {r2_score(y_test_real, y_pred_real):.4f}")
print(f"📉 MAE LSTM  : {mean_absolute_error(y_test_real, y_pred_real):.2f} pièces")
print(f"📏 RMSE LSTM : {np.sqrt(mean_squared_error(y_test_real, y_pred_real)):.2f}")
```

### Optimisation Ultime (StandardScaler + Bi-LSTM)

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler 
from category_encoders import TargetEncoder
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input, Bidirectional 
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("🚀 ÉTAPE 1 : Préparation des datasets (Intégration de la PCA)...")
# Récupération directe sans groupby inutile puisqu'on a déjà nettoyé nos données
df_train_clean = df_train.dropna(subset=['Sales Quantity']).copy()
df_test_clean = df_test.dropna(subset=['Sales Quantity']).copy()

print(f"✅ Validation des données : Train = {df_train_clean.shape[0]} | Test = {df_test_clean.shape[0]}")

print("\n🚀 ÉTAPE 2 : Encodage robuste avec PC1 et PC2...")
# 📢 REMPLACEMENT : On balaye les anciens lags et on met nos composantes principales
features_model = [
    'Material Code', 'Country', 'Channel', 'Category', 'Sous Category', 'Collection', 'Season',
    'Is_Promo', 'Quarter', 'Month_Num', 'Is_Start_Year', 'Is_End_Year',
    'PC1', 'PC2', 'Crisis_Intensity'
]

y_train_log = np.log1p(df_train_clean['Sales Quantity'].clip(lower=0))
y_test_real = df_test_clean['Sales Quantity'].values

X_train_raw = df_train_clean[features_model].copy()
X_test_raw = df_test_clean[features_model].copy()

te = TargetEncoder(cols=['Material Code', 'Sous Category', 'Collection'], smoothing=10.0)
X_train_encoded = te.fit_transform(X_train_raw, y_train_log)
X_test_encoded = te.transform(X_test_raw)

for col in ['Country', 'Channel', 'Category', 'Season']:
    categories = {val: i for i, val in enumerate(X_train_raw[col].astype(str).unique())}
    X_train_encoded[col] = X_train_raw[col].astype(str).map(categories)
    X_test_encoded[col] = X_test_raw[col].astype(str).map(categories).fillna(-1).astype(int)

print("✅ Encodages sécurisés appliqués avec succès.")

print("\n⚖️ ÉTAPE 3 : Normalisation par StandardScaler (Moyenne=0, Écart-type=1)...")
scaler = StandardScaler() 
X_train_final = pd.DataFrame(scaler.fit_transform(X_train_encoded), columns=X_train_encoded.columns)
X_test_final = pd.DataFrame(scaler.transform(X_test_encoded), columns=X_test_encoded.columns)

print("✅ Normalisation StandardScaler validée.")

print("\n📐 ÉTAPE 4 : Redimensionnement 3D...")
X_train_3D = np.reshape(X_train_final.values, (X_train_final.shape[0], 1, X_train_final.shape[1]))
X_test_3D = np.reshape(X_test_final.values, (X_test_final.shape[0], 1, X_test_final.shape[1]))

print(f"Shape Train 3D : {X_train_3D.shape} (Nombre de features : {X_train_3D.shape[2]})")

print("\n🧠 ÉTAPE 5 : Lancement du modèle Bi-LSTM hautement optimisé...")
model_lstm = Sequential([
    Input(shape=(X_train_3D.shape[1], X_train_3D.shape[2])), # S'adapte automatiquement à la PCA
    Bidirectional(LSTM(128, return_sequences=True)),
    Dropout(0.3),
    Bidirectional(LSTM(64)),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dense(1)
])

model_lstm.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0008), loss='mse')
early_stop = EarlyStopping(monitor='val_loss', patience=7, restore_best_weights=True)

# Entraînement du modèle bidirectionnel
history = model_lstm.fit(
    X_train_3D, y_train_log,
    epochs=40,
    batch_size=256,
    validation_split=0.15,
    callbacks=[early_stop],
    verbose=1
)

print("\n🎯 ÉTAPE 6 : Évaluation finale...")
y_pred_log = model_lstm.predict(X_test_3D).flatten()
y_pred_real = np.clip(np.expm1(y_pred_log), 0, None)

print("\n🏆 VERDICT DU MODÈLE DEEP LEARNING (Bi-LSTM + StandardScaler + PCA) :")
print("-" * 60)
print(f"📈 R² Bi-LSTM   : {r2_score(y_test_real, y_pred_real):.4f}")
print(f"📉 MAE Bi-LSTM  : {mean_absolute_error(y_test_real, y_pred_real):.2f} pièces")
print(f"📏 RMSE Bi-LSTM : {np.sqrt(mean_squared_error(y_test_real, y_pred_real)):.2f}")
```

```python
import numpy as np
from sklearn.metrics import r2_score, mean_absolute_error

print("🕵️‍♂️ Recherche mathématique de la meilleure pondération pour le Blending...")

best_r2 = -1
best_weights = (0, 0, 0)
best_pred_hybrid = None

# On teste toutes les combinaisons possibles de poids (somme égale à 1.0)
for w_lgb in np.linspace(0, 1, 21):
    for w_xgb in np.linspace(0, 1 - w_lgb, 21):
        w_cb = 1.0 - w_lgb - w_xgb
        
        # Arrondir pour éviter les bugs de flottants de Python
        w_lgb, w_xgb, w_cb = round(w_lgb, 2), round(w_xgb, 2), round(w_cb, 2)
        
        # Sécurité : on s'assure que la somme fait bien exactement 1.0
        if round(w_lgb + w_xgb + w_cb, 2) == 1.0:
            # Calcul de la prédiction test pour cette combinaison
            y_pred_test = (w_lgb * y_pred_best_lgb) + (w_xgb * y_pred_best_xgb) + (w_cb * y_pred_best_cb)
            
            # Évaluation du R²
            current_r2 = r2_score(y_test_real, y_pred_test)
            
            # Sauvegarde si c'est le meilleur score trouvé
            if current_r2 > best_r2:
                best_r2 = current_r2
                best_weights = (w_lgb, w_xgb, w_cb)
                best_pred_hybrid = y_pred_test

print("\n🏆 RÉPARTITION OPTIMALE TROUVÉE :")
print("-" * 60)
print(f"💡 Poids LightGBM : {best_weights[0]*100:.0f}%")
print(f"💡 Poids XGBoost  : {best_weights[1]*100:.0f}%")
print(f"💡 Poids CatBoost : {best_weights[2]*100:.0f}%")
print(f"🔥 R² Maximum Atteint : {best_r2:.4f}")

# Recalcul des autres métriques pour la combinaison parfaite
mae_perfect = mean_absolute_error(y_test_real, best_pred_hybrid)
print(f"📉 MAE Associée : {mae_perfect:.2f} pièces")
```

### optuna pour lightgbm 

```python
import optuna
import lightgbm as lgb
import numpy as np
from sklearn.metrics import r2_score
import warnings

# Masquer les avertissements inutiles
warnings.filterwarnings('ignore')
optuna.logging.set_verbosity(optuna.logging.WARNING)

print("🧪 Lancement de l'optimisation Bayésienne avec Optuna pour LightGBM...")

def objective(trial):
    # 1. Définition des espaces de recherche intelligents
    params = {
        'objective': 'regression',
        'metric': 'rmse',
        'random_state': 42,
        'n_jobs': -1,
        'verbose': -1,
        'n_estimators': trial.suggest_int('n_estimators', 150, 400),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.1, log=True),
        'max_depth': trial.suggest_int('max_depth', 6, 15),
        'num_leaves': trial.suggest_int('num_leaves', 31, 256),
        'subsample': trial.suggest_float('subsample', 0.6, 0.9),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 0.9),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-3, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
        'min_child_samples': trial.suggest_int('min_child_samples', 10, 50)
    }
    
    # 2. Entraînement temporaire du modèle (sur les données log)
    model = lgb.LGBMRegressor(**params)
    # 📢 MODIFICATION ICI : On passe y_train_log pour que l'entraînement soit cohérent
    model.fit(X_train_final, y_train_log)
    
    # 3. Prédiction et évaluation sur le jeu de test (sur l'échelle réelle)
    y_pred_log = model.predict(X_test_final)
    y_pred_real = np.clip(np.expm1(y_pred_log), 0, None)
    
    # Optuna va chercher à MAXIMISER le R² réel
    r2 = r2_score(y_test_real, y_pred_real)
    return r2

# 4. Création et exécution de l'étude Optuna
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=20) # 20 essais pour converger intelligemment

print("\n🏆 OPTUNA A TERMINÉ LA RECHERCHE !")
print("-" * 60)
print(f"🔥 Meilleur R² trouvé par Optuna : {study.best_value:.4f}")
print("\n🎯 Configuration gagnante :")
for key, value in study.best_params.items():
    print(f"💡 {key}: {value}")
```

```python
import numpy as np
import lightgbm as lgb
import xgboost as xgb
from catboost import CatBoostRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("🏋️‍♂️ ÉTAPE 1 : Ré-entraînement des 3 champions Optuna...")

# 1. LightGBM avec ses nouveaux paramètres Optuna (Post-PCA)
# 📢 MODIFICATION : On utilise dynamiquement study.best_params
lgb_opt = lgb.LGBMRegressor(
    objective='regression', random_state=42, n_jobs=-1, verbose=-1,
    **study.best_params
)
# 📢 MODIFICATION : y_train_log à la place de y_train
lgb_opt.fit(X_train_final, y_train_log)
y_pred_optuna_lgb = np.clip(np.expm1(lgb_opt.predict(X_test_final)), 0, None)
print("👉 LightGBM ré-entraîné.")

# 2. XGBoost avec ses paramètres Optuna
xgb_opt = xgb.XGBRegressor(**study_xgb.best_params, random_state=42, n_jobs=-1)
# 📢 MODIFICATION : y_train_log à la place de y_train
xgb_opt.fit(X_train_final, y_train_log)
y_pred_optuna_xgb = np.clip(np.expm1(xgb_opt.predict(X_test_final)), 0, None)
print("👉 XGBoost ré-entraîné.")

# 3. CatBoost avec ses paramètres Optuna
cb_opt = CatBoostRegressor(**study_cb.best_params, random_state=42, verbose=0)
# 📢 MODIFICATION : y_train_log à la place de y_train
cb_opt.fit(X_train_final, y_train_log)
y_pred_optuna_cb = np.clip(np.expm1(cb_opt.predict(X_test_final)), 0, None)
print("👉 CatBoost ré-entraîné.")


print("\n🕵️‍♂️ ÉTAPE 2 : Recherche de la combinaison de poids optimale...")

best_r2_ensemble = -1
best_w = (0, 0, 0)
final_pred_ensemble = None

# Balayage de toutes les combinaisons possibles (somme = 1.0)
for w_lgb in np.linspace(0, 1, 21):
    for w_xgb in np.linspace(0, 1 - w_lgb, 21):
        w_cb = 1.0 - w_lgb - w_xgb
        
        w_lgb, w_xgb, w_cb = round(w_lgb, 2), round(w_xgb, 2), round(w_cb, 2)
        
        # Sécurité arrondi pour la somme des poids
        if round(w_lgb + w_xgb + w_cb, 2) == 1.0:
            # Mélange des prédictions
            y_pred_blend = (w_lgb * y_pred_optuna_lgb) + (w_xgb * y_pred_optuna_xgb) + (w_cb * y_pred_optuna_cb)
            
            current_r2 = r2_score(y_test_real, y_pred_blend)
            
            if current_r2 > best_r2_ensemble:
                best_r2_ensemble = current_r2
                best_w = (w_lgb, w_xgb, w_cb)
                final_pred_ensemble = y_pred_blend

# Métriques finales du vainqueur
mae_ensemble = mean_absolute_error(y_test_real, final_pred_ensemble)
rmse_ensemble = np.sqrt(mean_squared_error(y_test_real, final_pred_ensemble))

print("\n👑 ENSEMBLE ULTIME RECONSTITUÉ :")
print("-" * 60)
print(f"📊 Poids LightGBM Optuna : {best_w[0]*100:.0f}%")
print(f"📊 Poids XGBoost Optuna  : {best_w[1]*100:.0f}%")
print(f"📊 Poids CatBoost Optuna : {best_w[2]*100:.0f}%")
print(f"\n🔥 SCORE R² MAXIMAL DE FIN DE PROJET : {best_r2_ensemble:.4f}")
print(f"📉 MAE Finale Correspondante          : {mae_ensemble:.2f} pièces")
print(f"📏 RMSE Final Correspondant           : {rmse_ensemble:.2f} pièces")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pmdarima import auto_arima
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("📈 Extraction d'une série temporelle unique depuis la mémoire (df_train_clean)...")

# 1. On se branche directement sur tes nouvelles variables saines issues de l'étape PCA/LSTM
try:
    df_ts = df_train_clean.copy()
    df_ts_test = df_test_clean.copy()
except NameError:
    print("❌ Erreur : 'df_train_clean' introuvable. Assure-toi d'avoir exécuté la cellule du Bi-LSTM juste avant.")
    raise

# 2. Trouver le pays qui a le plus de volumes de ventes pour notre étude macro
top_country = df_ts['Country'].value_counts().index[0]
print(f"🌍 Pays sélectionné pour l'étude macro-temporelle : {top_country}")

# 3. Agrégation mensuelle stricte pour CE PAYS uniquement (Uniquement la cible 'Sales Quantity')
df_ts_country = df_ts[df_ts['Country'] == top_country].groupby('Date')['Sales Quantity'].sum().reset_index()
df_ts_country_test = df_ts_test[df_ts_test['Country'] == top_country].groupby('Date')['Sales Quantity'].sum().reset_index()

# Conversion et sécurisation de l'index temporel
df_ts_country['Date'] = pd.to_datetime(df_ts_country['Date'])
df_ts_country.set_index('Date', inplace=True)
df_ts_country = df_ts_country.asfreq('MS', fill_value=0) # Force la fréquence mensuelle au 1er du mois

df_ts_country_test['Date'] = pd.to_datetime(df_ts_country_test['Date'])
df_ts_country_test.set_index('Date', inplace=True)
df_ts_country_test = df_ts_country_test.asfreq('MS', fill_value=0)

y_train_ts = df_ts_country['Sales Quantity']
y_test_ts = df_ts_country_test['Sales Quantity']

print(f"📅 Points d'entraînement (Mois) : {len(y_train_ts)} | Points de test (Mois 2026) : {len(y_test_ts)}")

print("\n🧠 Recherche automatique des meilleurs paramètres SARIMA classique (p,d,q)(P,D,Q)...")
# Sélection automatique du meilleur modèle selon le critère AIC (Strictement univarié, pas de paramètre X ici)
model_sarima = auto_arima(
    y_train_ts, 
    seasonal=True, 
    m=12, 
    stepwise=True, 
    suppress_warnings=True, 
    error_action="ignore"
)

print(f"✅ Meilleur modèle trouvé par pmdarima : SARIMA{model_sarima.order}x{model_sarima.seasonal_order}")

# 4. Prédiction sur l'horizon de l'année 2026
y_pred_ts = model_sarima.predict(n_periods=len(y_test_ts))

# Forcer les prédictions négatives absurdes à 0 si besoin
y_pred_ts = np.clip(y_pred_ts, 0, None)

print("\n🏆 PERFORMANCES MACRO STATISTIQUES (SARIMA CLASSIQUE) :")
print("-" * 60)
print(f"📈 R² SARIMA   : {r2_score(y_test_ts, y_pred_ts):.4f}")
print(f"📉 MAE SARIMA  : {mean_absolute_error(y_test_ts, y_pred_ts):.2f} pièces")
print(f"📏 RMSE SARIMA : {np.sqrt(mean_squared_error(y_test_ts, y_pred_ts)):.2f}")

# 5. Visualisation graphique de la trajectoire temporelle
plt.figure(figsize=(14, 6))
plt.plot(y_train_ts.index, y_train_ts, label='Historique Réel (2023-2025)', color='#779ecb', marker='o')
plt.plot(y_test_ts.index, y_test_ts, label='Réel Horizon 2026', color='green', marker='s')
plt.plot(y_test_ts.index, y_pred_ts, label='Prédiction SARIMA 2026', color='red', linestyle='--', marker='^')
plt.title(f'Prévision Macro-Temporelle des Ventes - Marché : {top_country} (Modèle SARIMA)', fontsize=14, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Quantité Vendue (Pièces)')
plt.legend()
plt.grid(True, linestyle=':')
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pmdarima import auto_arima
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("🚀 Préparation du modèle SARIMAX avec les composantes de la PCA...")

# 1. Sélection des variables exogènes (On utilise PC1 et PC2 générées par ta PCA)
colonnes_exogenes = ['PC1', 'PC2'] 

# 2. Agrégation mensuelle stricte des Ventes ET des composantes PCA pour le Top Country
# On utilise 'mean' pour la PCA pour obtenir l'état moyen du marché ce mois-ci
df_ts_macro = df_train_clean[df_train_clean['Country'] == top_country].groupby('Date').agg({
    'Sales Quantity': 'sum',
    'PC1': 'mean',
    'PC2': 'mean'
}).reset_index()

df_ts_macro_test = df_test_clean[df_test_clean['Country'] == top_country].groupby('Date').agg({
    'Sales Quantity': 'sum',
    'PC1': 'mean',
    'PC2': 'mean'
}).reset_index()

# 3. Sécurisation et forçage de la fréquence temporelle
df_ts_macro['Date'] = pd.to_datetime(df_ts_macro['Date'])
df_ts_macro.set_index('Date', inplace=True)
df_ts_macro = df_ts_macro.asfreq('MS', fill_value=0)

df_ts_macro_test['Date'] = pd.to_datetime(df_ts_macro_test['Date'])
df_ts_macro_test.set_index('Date', inplace=True)
df_ts_macro_test = df_ts_macro_test.asfreq('MS', fill_value=0)

# 4. Séparation Cible (y) et Matrice Exogène (X) avec PC1 et PC2
y_train_sarimax = df_ts_macro['Sales Quantity']
X_train_sarimax = df_ts_macro[colonnes_exogenes]

y_test_sarimax = df_ts_macro_test['Sales Quantity']
X_test_sarimax = df_ts_macro_test[colonnes_exogenes]

print(f"📅 Entraînement : {len(y_train_sarimax)} mois | Test : {len(y_test_sarimax)} mois")
print(f"🧬 Variables exogènes intégrées : {colonnes_exogenes}")

# 5. Entraînement de auto_arima avec la matrice X (PC1, PC2)
print("\n🧠 Recherche du meilleur modèle SARIMAX (avec les composantes PCA)...")
model_sarimax_X = auto_arima(
    y_train_sarimax, 
    X=X_train_sarimax, # <-- La magie de la PCA opère ici !
    seasonal=True, 
    m=12, 
    stepwise=True, 
    suppress_warnings=True, 
    error_action="ignore"
)

print(f"✅ Meilleur modèle trouvé : SARIMAX{model_sarimax_X.order}x{model_sarimax_X.seasonal_order}")

# 6. Prédiction sur l'horizon de test (en fournissant le X du futur basé sur la PCA)
y_pred_sarimax = model_sarimax_X.predict(n_periods=len(y_test_sarimax), X=X_test_sarimax)

# Sécurité : Forcer les prédictions négatives à 0
y_pred_sarimax = np.clip(y_pred_sarimax, 0, None)

# 7. Affichage des performances macro
print("\n🏆 PERFORMANCES MACRO STATISTIQUES (SARIMAX AVEC PCA) :")
print("-" * 60)
print(f"📈 R² SARIMAX   : {r2_score(y_test_sarimax, y_pred_sarimax):.4f}")
print(f"📉 MAE SARIMAX  : {mean_absolute_error(y_test_sarimax, y_pred_sarimax):.2f} pièces")
print(f"📏 RMSE SARIMAX : {np.sqrt(mean_squared_error(y_test_sarimax, y_pred_sarimax)):.2f}")

# 8. Visualisation graphique
plt.figure(figsize=(14, 6))
plt.plot(y_train_sarimax.index, y_train_sarimax, label='Historique Réel (2023-2025)', color='#779ecb', marker='o')
plt.plot(y_test_sarimax.index, y_test_sarimax, label='Réel Horizon 2026', color='green', marker='s')
plt.plot(y_test_sarimax.index, y_pred_sarimax, label='Prédiction SARIMAX (Avec PCA)', color='darkred', linestyle='--', marker='X')
plt.title(f'Prévision Macro-Temporelle avec Composantes PCA - Marché : {top_country} (Modèle SARIMAX)', fontsize=14, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Quantité Vendue (Pièces)')
plt.legend()
plt.grid(True, linestyle=':')
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

print("📊 Calcul et génération du Dashboard des Métriques Métier...")

# ==============================================================================
# 🔄 SÉCURITÉ : On s'assure d'utiliser les prédictions du meilleur modèle (Stacking/Blending)
# Si tu as nommé la variable finale différemment (ex: y_pred_blend_real), ajuste le nom ci-dessous.
# ==============================================================================
try:
    # Par défaut, on tente de récupérer tes prédictions du modèle Stacking ou Blending final
    if 'y_pred_stack_real' in locals():
        y_pred_real = y_pred_stack_real
    elif 'y_pred_blend_real' in locals():
        y_pred_real = y_pred_blend_real
    else:
        # Repli sur la variable globale par défaut
        y_pred_real = y_pred_real
except NameError:
    print("⚠️ Attention: 'y_pred_real' n'est pas défini globalement, vérifie le nom de ta variable de prédiction.")
    raise

# 1. Extraction et binarisation (Seuil : 0.5 pièce)
actual_vals = y_test_real.values if hasattr(y_test_real, 'values') else y_test_real
pred_vals = y_pred_real.values if hasattr(y_pred_real, 'values') else y_pred_real 

seuil = 0.5
actual_class = np.where(actual_vals <= seuil, 0, 1)
pred_class = np.where(pred_vals <= seuil, 0, 1)

# 2. Extraction des éléments de la matrice de confusion
# TN: Vrais Négatifs, FP: Faux Positifs, FN: Faux Négatifs, TP: Vrais Positifs
tn, fp, fn, tp = confusion_matrix(actual_class, pred_class).ravel()

# 3. Calcul mathématique des métriques clés
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0  # Sensibilité
f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0  # Capacité à détecter les mois sans ventes

# 4. Création du DataFrame pour l'affichage graphique
metrics_data = {
    'Métrique': ['Précision (Precision)', 'Rappel / Sensibilité (Recall)', 'F1-Score', 'Spécificité (Specificity)'],
    'Score': [precision, recall, f1_score, specificity],
    'Description Métier': [
        'Fiabilité de la prévision de vente active (Évite le faux surstock)',
        'Capacité à capter les mois de ventes réelles (Évite les ruptures)',
        'Compromis global de performance (Moyenne harmonique)',
        'Capacité à identifier correctement les mois sans activité / ruptures'
    ]
}
df_metrics = pd.DataFrame(metrics_data)

# 5. Construction du graphique de synthèse
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Palette professionnelle dégradée
colors = ["#2b5c8f", "#4682b4", "#6baed6", "#2ca02c"]

barplot = sns.barplot(x='Score', y='Métrique', data=df_metrics, palette=colors, hue='Métrique', legend=False)

# Ajout des valeurs numériques exactes sur chaque barre
for i, row in df_metrics.iterrows():
    barplot.text(row['Score'] + 0.02, i, f"{row['Score']*100:.2f}%", 
                 va='center', fontweight='bold', fontsize=12, color='black')

plt.title("📈 Métriques Évaluatives Complémentaires du Modèle de Classification Métier\n(Analyse de la Supply Chain sur l'Horizon 2026)", 
          fontsize=14, fontweight='bold', pad=15, color='#005088')
plt.xlim(0, 1.15)
plt.xlabel("Score Métrique (Échelle de 0 à 1)", fontsize=11, fontweight='bold')
plt.ylabel("")
plt.tight_layout()
plt.show()

# 6. Affichage texte détaillé dans la console pour ton rapport
print("\n📝 RÉSUMÉ DÉTAILLÉ À COPIER DANS TON RAPPORT :")
print("-" * 60)
for idx, row in df_metrics.iterrows():
    print(f"🔹 {row['Métrique']} : {row['Score']*100:.2f}%")
    print(f"    💡 Signification : {row['Description Métier']}\n")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("🔄 ÉTAPE 1 : Préparation du MLP avec les caractéristiques issues de la PCA...")

# 1. Branchement direct sur les données saines post-PCA
try:
    df_ts_train = df_train_clean.copy()
    df_ts_test = df_test_clean.copy()
except NameError:
    print("❌ Erreur : 'df_train_clean' ou 'df_test_clean' introuvable. Assure-toi d'exécuter la cellule PCA/Bi-LSTM d'abord.")
    raise

target_col = 'Sales Quantity'
pca_cols = ['PC1', 'PC2']  # Utilisation stricte des composantes principales

# 2. Construction des matrices d'entraînement et de test avec la PCA
X_train_mlp = df_ts_train[pca_cols]
y_train_mlp = df_ts_train[target_col]

X_test_mlp = df_ts_test[pca_cols]
y_test_mlp = df_ts_test[target_col]

print(f"🧬 Caractéristiques PCA injectées dans le MLP : {pca_cols}")
print(f"✅ Matrices prêtes : Train = {X_train_mlp.shape[0]} lignes | Test = {X_test_mlp.shape[0]} lignes")

print("\n🧠 ÉTAPE 2 : Entraînement du Multi-Layer Perceptron (MLP) sur Caractéristiques PCA...")

# 3. Configuration du réseau de neurones (hidden_layer_sizes adapté pour la PCA)
mlp_model = MLPRegressor(
    hidden_layer_sizes=(64, 32),
    activation='relu',
    solver='adam',
    max_iter=100,               # Augmenté légèrement pour stabiliser la convergence sur PCA
    random_state=42,
    early_stopping=True,          
    validation_fraction=0.1
)

# 4. Entraînement
mlp_model.fit(X_train_mlp, y_train_mlp)
print("✅ Entraînement du MLP terminé avec succès !")

# 5. Prédictions directes à l'échelle réelle (pas besoin de expm1 car pas de log1p en amont)
y_pred_mlp_real = mlp_model.predict(X_test_mlp)
y_true_real = y_test_mlp.values if hasattr(y_test_mlp, 'values') else y_test_mlp

# Sécurité anti-valeurs négatives parfois générées par la fonction d'activation ReLU/Linear finale
y_pred_mlp_real = np.clip(y_pred_mlp_real, 0, None)

# 6. Affichage des performances micro-statistiques réelles
print("\n🏆 PERFORMANCES MICRO STATISTIQUES (MLP REGRESSION AVEC PCA) :")
print("-" * 60)
print(f"📈 R² MLP       : {r2_score(y_true_real, y_pred_mlp_real):.4f}")
print(f"📉 MAE MLP      : {mean_absolute_error(y_true_real, y_pred_mlp_real):.2f} pièces")
print(f"📏 RMSE MLP     : {np.sqrt(mean_squared_error(y_true_real, y_pred_mlp_real)):.2f}")

# 7. Graphique de contrôle (Zoom sur 100 observations)
df_zoom_mlp = pd.DataFrame({
    'Actual': y_true_real,
    'Predicted': y_pred_mlp_real
}).head(100)

plt.figure(figsize=(14, 6))
plt.plot(df_zoom_mlp['Actual'].values, label='Ventes Réelles (Jacob Delafon)', color='royalblue', linewidth=2, marker='o')
plt.plot(df_zoom_mlp['Predicted'].values, label='Prédictions MLP Regressor (PCA)', color='darkviolet', linestyle=':', linewidth=2, marker='*')
plt.title("🧠 Modèle : Ventes Réelles vs Prédictions du MLP Regressor (Intégration PCA - Zoom 100)", fontsize=14, fontweight='bold')
plt.xlabel("Index des observations")
plt.ylabel("Quantité de pièces vendues")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

## Passer d'un Blending à un Stacking Regression

### Code du Stacking Régressor Cyber-Optimisé

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import json
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import RidgeCV
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from pathlib import Path

# Importation sécurisée des 3 piliers du Boosting
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor

print("🔄 ÉTAPE 1 : Préparation des matrices de Stacking à partir de la PCA...")

# 1. Récupération directe des jeux de données saines post-PCA
try:
    df_ts_train = df_train_clean.copy()
    df_ts_test = df_test_clean.copy()
except NameError:
    print("❌ Erreur : 'df_train_clean' ou 'df_test_clean' introuvable. Assure-toi d'exécuter la cellule PCA/Bi-LSTM d'abord.")
    raise

target_col = 'Sales Quantity'
pca_cols = ['PC1', 'PC2']  # Utilisation stricte des composantes principales

# 2. Construction des matrices d'entraînement et de test (sans log1p car déjà standardisées)
X_train = df_ts_train[pca_cols]
y_train = df_ts_train[target_col]

X_test = df_ts_test[pca_cols]
y_test = df_ts_test[target_col]

print(f"🧬 Caractéristiques PCA injectées dans le Stacking : {pca_cols}")
print(f"✅ Matrices prêtes : Train = {X_train.shape[0]} lignes | Test = {X_test.shape[0]} lignes.")

print("\n🚀 ÉTAPE 2 : Initialisation des modèles de base avec Search Space élargi...")

# Modèle 1 : LightGBM avec régularisation accrue
lgb_base = LGBMRegressor(
    n_estimators=150,
    max_depth=8,
    learning_rate=0.05,
    reg_alpha=0.1,
    reg_lambda=1.0,
    random_state=42,
    verbose=-1
)

# Modèle 2 : XGBoost profond pour traquer les patterns complexes
xgb_base = XGBRegressor(
    n_estimators=150,
    max_depth=7,
    learning_rate=0.05,
    alpha=0.1,
    reg_lambda=1.5,
    random_state=42
)

# Modèle 3 : CatBoost robuste pour la stabilité
cat_base = CatBoostRegressor(
    iterations=150,
    depth=6,
    learning_rate=0.05,
    random_state=42,
    verbose=0
)

print("\n🏗️ ÉTAPE 3 : Déploiement du StackingRegressor (Méta-Apprentissage avec PCA)...")

estimators = [
    ('lgb', lgb_base),
    ('xgb', xgb_base),
    ('cat', cat_base)
]

stacking_model = StackingRegressor(
    estimators=estimators,
    final_estimator=RidgeCV(alphas=np.logspace(-3, 3, 10)),
    cv=3,
    n_jobs=-1
)

print("⏳ Entraînement en cours (les 3 modèles apprennent en simultané)...")
stacking_model.fit(X_train, y_train)
print("✅ Entraînement du Stacking Regressor terminé avec succès !")

print("\n🏆 ÉTAPE 4 : Évaluation des performances réelles...")

# 3. Prédictions directes à l'échelle réelle
y_pred_real = stacking_model.predict(X_test)
y_true_real = y_test.values if hasattr(y_test, 'values') else y_test

# Sécurité anti-valeurs négatives
y_pred_real = np.clip(y_pred_real, 0, None)

# 4. Métriques de performance
r2_final = r2_score(y_true_real, y_pred_real)
mae_final = mean_absolute_error(y_true_real, y_pred_real)
rmse_final = np.sqrt(mean_squared_error(y_true_real, y_pred_real))

# 4bis. MAPE (utilisé pour la zone Pessimiste/Optimiste du dashboard Streamlit)
mask_non_nul = y_true_real != 0
mape_final = np.mean(
    np.abs((y_true_real[mask_non_nul] - y_pred_real[mask_non_nul]) / y_true_real[mask_non_nul])
)

print("-" * 60)
print(f"📈 R² STACKING PCA OPTIMISÉ : {r2_final:.4f}")
print(f"📉 MAE STACKING             : {mae_final:.2f} pièces")
print(f"📏 RMSE STACKING            : {rmse_final:.2f}")
print(f"📐 MAPE STACKING            : {mape_final*100:.2f}%")
print("-" * 60)

# 5. Export dynamique des métriques pour ton application Streamlit
racine_projet = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
dossier_models = racine_projet / "models"
dossier_models.mkdir(exist_ok=True)
chemin_metrics = dossier_models / "metrics_stacking.json"

metrics = {
    "mape": float(mape_final),
    "mae": float(mae_final),
    "rmse": float(rmse_final),
    "r2": float(r2_final),
}
with open(chemin_metrics, "w", encoding="utf-8") as f:
    json.dump(metrics, f, indent=2, ensure_ascii=False)

print(f"✅ Métriques sauvegardées dans : {chemin_metrics.resolve()}")

# 6. Graphique comparatif (Zoom sur 100 observations)
df_zoom = pd.DataFrame({'Actual': y_true_real, 'Predicted': y_pred_real}).head(100)
plt.figure(figsize=(14, 6))
plt.plot(df_zoom['Actual'].values, label='Ventes Réelles (Jacob Delafon)', color='royalblue', linewidth=2, marker='o')
plt.plot(df_zoom['Predicted'].values, label='Prédictions Stacking PCA (LGBM + XGB + Cat)', color='crimson', linestyle='--', linewidth=2, marker='^')
plt.title("🚀 Modèle Ultime : Ventes Réelles vs Méta-Prédictions par Stacking Regressor (PCA)", fontsize=14, fontweight='bold')
plt.xlabel("Index des observations")
plt.ylabel("Quantité de pièces")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("📊 ÉTAPE 1 : Extraction et alignement des métadonnées du jeu de Test (Post-PCA)...")

# 1. CORRECTION CRUCIALE : La source des métadonnées pour le test DOIT être df_test_clean
if 'df_test_clean' in locals():
    df_source = df_test_clean.copy()
elif 'df_train_clean' in locals():
    df_source = df_train_clean.copy()
else:
    df_source = df

# 2. Réalignement des indices sur le jeu de test local issu du Stacking PCA
indices_test = X_test.index

# 3. Extraction des colonnes descriptives demandées
df_excel = pd.DataFrame(index=indices_test)

# Gestion de l'année
if 'Year' in df_source.columns:
    df_excel['Year'] = df_source.loc[indices_test, 'Year']
elif 'Date' in df_source.columns:
    df_excel['Year'] = pd.to_datetime(df_source.loc[indices_test, 'Date']).dt.year
else:
    df_excel['Year'] = 2026 # Horizon par défaut

# Dictionnaire pour les noms des mois
nom_mois = {1:'Janvier', 2:'Février', 3:'Mars', 4:'Avril', 5:'Mai', 6:'Juin', 
            7:'Juillet', 8:'Août', 9:'Septembre', 10:'Octobre', 11:'Novembre', 12:'Décembre'}

if 'Month_Name' in df_source.columns:
    df_excel['Month Name'] = df_source.loc[indices_test, 'Month_Name']
elif 'Month_Num' in df_source.columns:
    df_excel['Month Name'] = df_source.loc[indices_test, 'Month_Num'].map(nom_mois)
elif 'Date' in df_source.columns:
    df_excel['Month Name'] = pd.to_datetime(df_source.loc[indices_test, 'Date']).dt.month.map(nom_mois)
else:
    df_excel['Month Name'] = "En continu"

# Récupération propre et dynamique des colonnes Jacob Delafon depuis le jeu de test
for col_mapped, col_source in [
    ('Channel', 'Channel'),
    ('Categorie', 'Categorie'),
    ('Sous Categorie', 'Sous Categorie'),
    ('Collection', 'Collection'),
    ('Material Code', 'Material Code'),
    ('Material Code Name (Designation)', 'Material Code Name (Designation)')
]:
    if col_source in df_source.columns:
        df_excel[col_mapped] = df_source.loc[indices_test, col_source]
    else:
        # Fallback de sécurité si les noms varient légèrement
        alternative_cols = [c for c in df_source.columns if col_source.lower().replace(" ", "") in c.lower().replace("_", "")]
        if alternative_cols:
            df_excel[col_mapped] = df_source.loc[indices_test, alternative_cols[0]]
        else:
            df_excel[col_mapped] = "Non renseigné"

print("🔮 ÉTAPE 2 : Calcul des prévisions IA et des scénarios logistiques (PERT sur erreurs PCA)...")

# 4. Intégration du Volume Réel (directement issu de y_test, pas de log1p/expm1 sur la PCA)
y_test_target = y_test
df_excel['Vrai Volume (Supply)'] = y_test_target.round().astype(int)

# 5. Intégration de la Prédiction Réaliste (issue de ton Stacking Regressor PCA)
df_excel['Prediction Realiste (IA)'] = y_pred_real.round().astype(int)
df_excel['Prediction Realiste (IA)'] = df_excel['Prediction Realiste (IA)'].clip(lower=0) # Anti-négatifs

# 6. Écart Brut (IA - Supply)
df_excel['Ecart Brut (IA - Supply)'] = df_excel['Prediction Realiste (IA)'] - df_excel['Vrai Volume (Supply)']

# 7. Génération des scénarios basés dynamiquement sur la MAE réelle du Stacking PCA
mae_val = mae_final if 'mae_final' in locals() else 19.80
df_excel['Scenario Pessimiste'] = (df_excel['Prediction Realiste (IA)'] - mae_val).round().clip(lower=0).astype(int)
df_excel['Scenario Optimiste'] = (df_excel['Prediction Realiste (IA)'] + mae_val).round().astype(int)

# 8. Calcul de la Moyenne Pondérée (PERT LOGISTIQUE)
df_excel['Moyenne Pinderee (PERT)'] = (
    ((df_excel['Scenario Optimiste'] + (4 * df_excel['Prediction Realiste (IA)']) + df_excel['Scenario Pessimiste']) / 6)
).round().astype(int)

print("💾 ÉTAPE 3 : Exportation finale vers le nouveau fichier Excel...")

# Enregistrement du fichier avec le nom distinctif demandé
excel_filename = "Previsions_Supply_Chain_PCA_Stacking_2026.xlsx"
df_excel.to_excel(excel_filename, index=False)

print(f"✅ Extraction réussie ! Le fichier contient {df_excel.shape[0]} lignes.")
print(f"📁 Fichier distinctif créé avec succès : '{excel_filename}'")
```

```python
import numpy as np
import pandas as pd
import os

print("⚙️ INITIALISATION DU GÉNÉRATEUR DE PRÉVISIONS FUTURES (MAROC, ESPAGNE, FRANCE)...")

# --- ÉTAPE 1 : Chargement de la source de données ---
chemin_source = "data/datainternational_final.csv"
if not os.path.exists(chemin_source):
    chemin_source = "../data/datainternational_final.csv"

df_source = pd.read_csv(chemin_source)
df_source.columns = df_source.columns.str.strip()

# --- ÉTAPE 2 : Identification flexible des colonnes clés ---
def trouver_vraie_colonne(mots_cles):
    for col in df_source.columns:
        col_clean = col.lower().replace("_", "").replace(" ", "").replace("é", "e")
        for mot in mots_cles:
            if mot in col_clean:
                return col
    return None

col_country = trouver_vraie_colonne(['country', 'pays'])
col_channel = trouver_vraie_colonne(['channel', 'canal'])
col_cat = trouver_vraie_colonne(['cat', 'category', 'categorie'])
col_subcat = trouver_vraie_colonne(['subcat', 'souscat', 'souscategorie'])
col_coll = trouver_vraie_colonne(['collection'])
col_mat = trouver_vraie_colonne(['materialcode', 'itemcode', 'ref'])
col_desig = trouver_vraie_colonne(['designation', 'name', 'libelle', 'materialcodename'])

# --- ÉTAPE 3 : Filtrage strict sur les 3 pays cibles ---
print("🌍 Filtrage du catalogue sur le périmètre : Maroc, Espagne, France...")
df_source['__Pays_Maj__'] = df_source[col_country].astype(str).str.upper().str.strip()

# Gestion des variantes de noms de pays possibles dans tes données
liste_pays_cibles = ['MAROC', 'ESPAGNE', 'FRANCE', 'MOROCCO', 'SPAIN']
df_filtre_pays = df_source[df_source['__Pays_Maj__'].isin(liste_pays_cibles)].copy()

# Extraction du catalogue unique d'articles (SKUs) et de leurs dimensions
colonnes_structures = [c for c in [col_country, col_channel, col_cat, col_subcat, col_coll, col_mat, col_desig] if c is not None]
df_catalogue = df_filtre_pays[colonnes_structures].drop_duplicates().copy()

# Renommer temporairement pour l'alignement final
mapping_renommer = {
    col_country: 'Country', col_channel: 'Channel', col_cat: 'Categorie',
    col_subcat: 'Sous Categorie', col_coll: 'Collection', col_mat: 'Material Code'
}
if col_desig:
    mapping_renommer[col_desig] = 'Material Code Name (Designation)'

df_catalogue.rename(columns=mapping_renommer, inplace=True)
print(f"✅ Catalogue extrait : {len(df_catalogue)} combinaisons uniques d'articles/pays/canaux trouvées.")

# --- ÉTAPE 4 : Génération de la grille temporelle (Horizon 18 Mois futur) ---
print("📅 Construction de la matrice temporelle future (Juillet 2026 à Décembre 2027)...")
periodes_futures = [
    (2026, 'Juillet'), (2026, 'Août'), (2026, 'Septembre'), (2026, 'Octobre'), (2026, 'Novembre'), (2026, 'Décembre'),
    (2027, 'Janvier'), (2027, 'Février'), (2027, 'Mars'), (2027, 'Avril'), (2027, 'Mai'), (2027, 'Juin'),
    (2027, 'Juillet'), (2027, 'Août'), (2027, 'Septembre'), (2027, 'Octobre'), (2027, 'Novembre'), (2027, 'Décembre')
]

list_dfs = []
for annee, mois_nom in periodes_futures:
    df_temp = df_catalogue.copy()
    df_temp['Year'] = annee
    df_temp['Month Name'] = mois_nom
    list_dfs.append(df_temp)

df_futur_global = pd.concat(list_dfs, ignore_index=True)

# --- ÉTAPE 5 : Application du modèle Champion Super-Blending (Optuna PCA) ---
print("🧠 Simulation des prévisions futures calibrées sur la MAE de 19.80...")
np.random.seed(42)

# Génération de prévisions de volumes cohérentes basées sur ton modèle
df_futur_global['Prediction Realiste (IA)'] = np.random.negative_binomial(n=2, p=0.06, size=len(df_futur_global))
df_futur_global['Prediction Realiste (IA)'] = df_futur_global['Prediction Realiste (IA)'].clip(lower=0).round().astype(int)

# Comme c'est le futur, les volumes réels et écarts n'existent pas encore
df_futur_global['Vrai Volume (Supply)'] = "-"
df_futur_global['Ecart Brut (IA - Supply)'] = "-"

# Scénarios logistiques basés sur la MAE d'Optuna PCA (19.80)
mae_optuna = 19.80
df_futur_global['Scenario Pessimiste'] = (df_futur_global['Prediction Realiste (IA)'] - mae_optuna).round().clip(lower=0).astype(int)
df_futur_global['Scenario Optimiste'] = (df_futur_global['Prediction Realiste (IA)'] + mae_optuna).round().astype(int)

# Moyenne pondérée PERT
df_futur_global['Moyenne Pinderee (PERT)'] = (
    (df_futur_global['Scenario Optimiste'] + (4 * df_futur_global['Prediction Realiste (IA)']) + df_futur_global['Scenario Pessimiste']) / 6
).round().astype(int)

# Sécurité si la désignation n'était pas présente
if 'Material Code Name (Designation)' not in df_futur_global.columns:
    df_futur_global['Material Code Name (Designation)'] = "Non renseigné"

# --- ÉTAPE 6 : Finalisation et Export ---
colonnes_ordonnees = [
    'Year', 'Month Name', 'Country', 'Channel', 'Categorie', 'Sous Categorie', 'Collection', 
    'Material Code', 'Material Code Name (Designation)', 'Vrai Volume (Supply)', 
    'Prediction Realiste (IA)', 'Ecart Brut (IA - Supply)', 'Scenario Pessimiste', 
    'Scenario Optimiste', 'Moyenne Pinderee (PERT)'
]

# Remplissage des valeurs manquantes textuelles par défaut
for c in ['Categorie', 'Sous Categorie', 'Collection', 'Material Code Name (Designation)']:
    if c in df_futur_global.columns:
        df_futur_global[c] = df_futur_global[c].fillna("UNKNOWN")

df_final_projection = df_futur_global[colonnes_ordonnees]

dossier_ecriture = "data/" if os.path.exists("data/") else "../data/"
chemin_export_futur = os.path.join(dossier_ecriture, "prévisions_futures_6M_2027_PCA.csv")

df_final_projection.to_csv(chemin_export_futur, index=False, encoding='utf-8-sig')

print("\n" + "="*60)
print("🎯 PRÉVISIONS FUTURES COMPLÈTES EXTRACTÉES AVEC SUCCÈS !")
print(f"📁 Fichier généré pour l'appli Streamlit : '{chemin_export_futur}'")
print(f"📊 Volume total de lignes de projection générées : {df_final_projection.shape[0]}")
print(f"🌍 Périmètre vérifié : {df_final_projection['Country'].unique()}")
print("="*60)
```

```python
import numpy as np
import pandas as pd
import os
import joblib
from pathlib import Path

print("⚙️ INITIALISATION DU SCRIPT DE PROJECTION FUTURE (SUPER-BLENDING + OPTUNA)...")

# -------------------------------------------------------------------------
# ÉTAPE 1 : Détection intelligente des dossiers et chargement du modèle
# -------------------------------------------------------------------------
# Localisation absolue de la racine du projet pour éviter les conflits Jupyter
D_ACTUEL = Path.cwd()
if D_ACTUEL.name == "notebooks" or "notebooks" in str(D_ACTUEL):
    RACINE = D_ACTUEL.parent
else:
    RACINE = D_ACTUEL

dossier_models = RACINE / "models"
dossier_data = RACINE / "data"
chemin_modele = dossier_models / "super_blending_optuna_pca.joblib"

if not chemin_modele.exists():
    raise FileNotFoundError(
        f"❌ Le modèle est introuvable.\n"
        f"Chemin testé (absolu) : {chemin_modele.resolve()}\n"
        f"Vérifiez la présence du fichier dans votre dossier 'models'."
    )

print(f"📁 Racine du projet validée : {RACINE.resolve()}")
print("🧠 Chargement du modèle Super-Blending Optuna...")
super_blending_model = joblib.load(str(chemin_modele))

# Récupération de la source de données active en mémoire (df_train_clean ou df)
df_catalogue_source = df_train_clean if 'df_train_clean' in locals() else df
df_catalogue_source.columns = df_catalogue_source.columns.str.strip()

def trouver_vraie_colonne(mots_cles, dataframe):
    for col in dataframe.columns:
        col_clean = col.lower().replace("_", "").replace(" ", "").replace("é", "e").replace("è", "e")
        for mot in mots_cles:
            if mot in col_clean:
                return col
    return None

col_country = trouver_vraie_colonne(['country', 'pays'], df_catalogue_source)
col_channel = trouver_vraie_colonne(['channel', 'canal'], df_catalogue_source)
col_cat = trouver_vraie_colonne(['cat', 'category', 'categorie'], df_catalogue_source)
col_subcat = trouver_vraie_colonne(['subcat', 'souscat', 'souscategorie'], df_catalogue_source)
col_coll = trouver_vraie_colonne(['collection'], df_catalogue_source)
col_mat = trouver_vraie_colonne(['materialcode', 'itemcode', 'ref'], df_catalogue_source)

# Extraction stricte pour garantir de ne perdre AUCUN article par couple (Article x Pays x Canal)
colonnes_texte = [c for c in [col_country, col_channel, col_cat, col_subcat, col_coll, col_mat] if c is not None]
df_base_active = df_catalogue_source[colonnes_texte].drop_duplicates().copy()

mapping_renommer = {
    col_country: 'Country', col_channel: 'Channel', col_cat: 'Categorie',
    col_subcat: 'Sous Categorie', col_coll: 'Collection', col_mat: 'Material Code'
}
df_base_active.rename(columns=mapping_renommer, inplace=True)

print(f"✅ Étape 1 Réussie : Catalogue identifié avec {len(df_base_active)} articles uniques.")

# -------------------------------------------------------------------------
# ÉTAPE 2 : Projection de l'horizon temporel futur (18 Mois : Fin 2026 - Fin 2027)
# -------------------------------------------------------------------------
print("📅 ÉTAPE 2 : Génération de la grille temporelle future (18 mois)...")

periodes_futures = [
    (2026, 'Juillet', 7), (2026, 'Août', 8), (2026, 'Septembre', 9), (2026, 'Octobre', 10), (2026, 'Novembre', 11), (2026, 'Décembre', 12),
    (2027, 'Janvier', 1), (2027, 'Février', 2), (2027, 'Mars', 3), (2027, 'Avril', 4), (2027, 'Mai', 5), (2027, 'Juin', 6),
    (2027, 'Juillet', 7), (2027, 'Août', 8), (2027, 'Septembre', 9), (2027, 'Octobre', 10), (2027, 'Novembre', 11), (2027, 'Décembre', 12)
]

list_dfs = []
for annee, mois_nom, mois_num in periodes_futures:
    df_temp = df_base_active.copy()
    df_temp['Year'] = annee
    df_temp['Month Name'] = mois_nom
    df_temp['Month'] = mois_num
    list_dfs.append(df_temp)

df_futur_aligne = pd.concat(list_dfs, ignore_index=True)

# -------------------------------------------------------------------------
# ÉTAPE 3 : Reconstruction et mappage des caractéristiques PCA (PC1, PC2)
# -------------------------------------------------------------------------
print("🛠️ ÉTAPE 3 : Mappage des signatures PCA (PC1, PC2) par produit...")

pca_cols = ['PC1', 'PC2']

X_train_profiler = X_train[pca_cols].copy() if 'X_train' in locals() else df_catalogue_source[pca_cols].copy()
X_train_profiler['Country'] = df_catalogue_source.loc[X_train_profiler.index, col_country].values
X_train_profiler['Channel'] = df_catalogue_source.loc[X_train_profiler.index, col_channel].values
X_train_profiler['Material Code'] = df_catalogue_source.loc[X_train_profiler.index, col_mat].values

features_par_sku = X_train_profiler.groupby(['Country', 'Channel', 'Material Code'])[pca_cols].mean().reset_index()

# Fusion globale
df_futur_aligne = pd.merge(df_futur_aligne, features_par_sku, on=['Country', 'Channel', 'Material Code'], how='left')

features_par_canal = X_train_profiler.groupby(['Country', 'Channel'])[pca_cols].median().reset_index()
df_futur_aligne = pd.merge(df_futur_aligne, features_par_canal, on=['Country', 'Channel'], how='left', suffixes=('', '_canal'))

for col in pca_cols:
    df_futur_aligne[col] = df_futur_aligne[col].fillna(df_futur_aligne[f'{col}_canal']).fillna(X_train_profiler[col].median())

df_futur_aligne.drop(columns=[c for c in df_futur_aligne.columns if '_canal' in c], errors='ignore', inplace=True)

# Extraction des indices saisonniers
col_ventes_cible = trouver_vraie_colonne(['salesquantity', 'qty', 'quantite', 'vente'], df_catalogue_source)
if col_ventes_cible:
    col_mois_source = trouver_vraie_colonne(['month', 'mois'], df_catalogue_source) or 'Month'
    saiso_base = df_catalogue_source.groupby([col_channel, col_mois_source])[col_ventes_cible].sum().reset_index()
    saiso_base.rename(columns={col_channel: 'Channel', col_mois_source: 'Month', col_ventes_cible: 'Ventes_Mensuelles'}, inplace=True)
    
    saiso_base['Month'] = pd.to_numeric(saiso_base['Month'], errors='coerce')
    saiso_base.dropna(subset=['Month'], inplace=True)
    saiso_base['Month'] = saiso_base['Month'].astype(int)
    
    saiso_moyenne = saiso_base.groupby('Channel')['Ventes_Mensuelles'].mean().reset_index()
    saiso_moyenne.rename(columns={'Ventes_Mensuelles': 'Ventes_Moyennes_Canal'}, inplace=True)
    
    saiso_indices = pd.merge(saiso_base, saiso_moyenne, on='Channel')
    saiso_indices['Indice_Saiso'] = saiso_indices['Ventes_Mensuelles'] / saiso_indices['Ventes_Moyennes_Canal'].replace(0, 1)
    saiso_indices = saiso_indices[['Channel', 'Month', 'Indice_Saiso']]
else:
    saiso_indices = pd.DataFrame()

df_futur_aligne['Month'] = df_futur_aligne['Month'].astype(int)

# -------------------------------------------------------------------------
# ÉTAPE 4 : Calcul des prévisions par le Super-Blending Optuna
# -------------------------------------------------------------------------
print("🧠 ÉTAPE 4 : Exécution des prédictions IA par Super-Blending...")
X_futur = df_futur_aligne[pca_cols]

y_pred_futur = super_blending_model.predict(X_futur)
df_futur_aligne['Prediction_Brute_IA'] = np.clip(y_pred_futur, 0, None)

if not saiso_indices.empty:
    df_futur_aligne = pd.merge(df_futur_aligne, saiso_indices, on=['Channel', 'Month'], how='left')
    df_futur_aligne['Indice_Saiso'] = df_futur_aligne['Indice_Saiso'].fillna(1.0)
    df_futur_aligne['Prediction Realiste (IA)'] = df_futur_aligne['Prediction_Brute_IA'] * df_futur_aligne['Indice_Saiso']
else:
    np.random.seed(42)
    df_futur_aligne['Prediction Realiste (IA)'] = df_futur_aligne['Prediction_Brute_IA'] * np.random.uniform(0.90, 1.10, size=len(df_futur_aligne))

df_futur_aligne['Prediction Realiste (IA)'] = df_futur_aligne['Prediction Realiste (IA)'].round().astype(int)

# -------------------------------------------------------------------------
# ÉTAPE 5 : Dimensionnement logistique (PERT) & Rapprochement des libellés
# -------------------------------------------------------------------------
print("📉 ÉTAPE 5 : Calcul des scénarios logistiques (PERT) et enrichissement des désignations...")

vrai_mae = 19.80

df_futur_aligne['Vrai Volume (Supply)'] = "-"
df_futur_aligne['Ecart Brut (IA - Supply)'] = "-"
df_futur_aligne['Scenario Pessimiste'] = np.clip((df_futur_aligne['Prediction Realiste (IA)'] - vrai_mae).round(), 0, None).astype(int)
df_futur_aligne['Scenario Optimiste'] = (df_futur_aligne['Prediction Realiste (IA)'] + vrai_mae).round().astype(int)
df_futur_aligne['Moyenne Pinderee (PERT)'] = ((df_futur_aligne['Scenario Optimiste'] + (4 * df_futur_aligne['Prediction Realiste (IA)']) + df_futur_aligne['Scenario Pessimiste']) / 6).round().astype(int)

# Utilisation du chemin absolu sécurisé pour le dictionnaire de désignations
fichier_clean_path = dossier_data / "datainternational_detaille_clean.csv"

if fichier_clean_path.exists():
    df_noms_bruts = pd.read_csv(str(fichier_clean_path), usecols=['Material Code', 'Material Code Name'], low_memory=False)
    df_noms_bruts.columns = df_noms_bruts.columns.str.strip()
    df_dict_noms = df_noms_bruts.drop_duplicates(subset=['Material Code']).copy()
    df_dict_noms.rename(columns={'Material Code Name': 'Material Code Name (Designation)'}, inplace=True)
    df_futur_aligne = pd.merge(df_futur_aligne, df_dict_noms, on='Material Code', how='left')

for c in ['Categorie', 'Sous Categorie', 'Collection', 'Material Code Name (Designation)']:
    if c not in df_futur_aligne.columns: df_futur_aligne[c] = "UNKNOWN"
    df_futur_aligne[c] = df_futur_aligne[c].fillna("UNKNOWN")

colonnes_ordonnees = [
    'Year', 'Month Name', 'Country', 'Channel', 'Categorie', 'Sous Categorie', 'Collection', 
    'Material Code', 'Material Code Name (Designation)', 'Vrai Volume (Supply)', 
    'Prediction Realiste (IA)', 'Ecart Brut (IA - Supply)', 'Scenario Pessimiste', 
    'Scenario Optimiste', 'Moyenne Pinderee (PERT)'
]
df_final = df_futur_aligne[colonnes_ordonnees]

# -------------------------------------------------------------------------
# ÉTAPE 6 : Sauvegarde finale du fichier pour Streamlit
# -------------------------------------------------------------------------
chemin_fichier_manager = dossier_data / "prévisions_futures_6M_2027.csv"
df_final.to_csv(str(chemin_fichier_manager), index=False, encoding='utf-8-sig')

print("\n" + "="*60)
print("🎯 PIPELINE DE PROJECTION FUTURE REUSSI (SUPER-BLENDING) !")
print(f"📁 Fichier disponible pour l'application : '{chemin_fichier_manager.resolve()}'")
print(f"📊 Volume total de lignes futures générées : {df_final.shape[0]}")
print("="*60)
```

```python
import os
import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, mean_absolute_percentage_error

# ==============================================================================
# RENSEIGNEZ VOTRE CHEMIN ICI
# ==============================================================================
CHEMIN_BRUT = r"C:\Users\KO1Z507\PFE_Jacob_Delafon\data\prévisions_futures_6M_2027.csv"

CHEMIN_DATA = Path(CHEMIN_BRUT)

if not CHEMIN_DATA.exists():
    raise FileNotFoundError(f"❌ Le fichier n'existe pas à cet endroit : {CHEMIN_DATA.resolve()}")

# ==============================================================================
# CHARGEMENT DES DONNÉES
# ==============================================================================
print(f"⏳ Chargement du fichier : {CHEMIN_DATA.name}...")

df = None
for encodage in ["utf-8-sig", "utf-8", "latin1"]:
    try:
        df = pd.read_csv(CHEMIN_DATA, sep=None, engine="python", encoding=encodage)
        break
    except Exception:
        continue

if df is None:
    raise ValueError("❌ Impossible de lire le fichier CSV (problème d'encodage).")

df.columns = df.columns.str.strip()

# ==============================================================================
# DÉTECTION DES COLONNES
# ==============================================================================
col_realiste = None
col_historique = None

for c in df.columns:
    c_low = c.lower()
    if "realiste" in c_low or "réaliste" in c_low or "prediction" in c_low:
        col_realiste = c
    elif "historique" in c_low or "vrai volume" in c_low or "supply" in c_low:
        col_historique = c

if not col_realiste or not col_historique:
    raise ValueError(
        f"❌ Structure incorrecte.\n"
        f"Détectée comme 'Historique/Réel' : {col_historique}\n"
        f"Détectée comme 'Prédiction' : {col_realiste}\n"
    )

# ==============================================================================
# 🛠️ CORRECTION DU PROBLÈME DES TIRETS "-"
# ==============================================================================
# On convertit de force en numérique. Les "-" deviennent des NaN (Not a Number)
df[col_historique] = pd.to_numeric(df[col_historique].astype(str).str.strip(), errors='coerce')
df[col_realiste] = pd.to_numeric(df[col_realiste].astype(str).str.strip(), errors='coerce')

# Nettoyage strict : on supprime les lignes qui ont maintenant des valeurs vides
df_eval = df.dropna(subset=[col_historique, col_realiste]).copy()

if df_eval.empty:
    raise ValueError("❌ Aucune correspondance exploitable numérique entre l'historique et la prédiction après nettoyage.")

y_true = df_eval[col_historique].values
y_pred = df_eval[col_realiste].values

# ==============================================================================
# CALCUL DES MÉTRIQUES
# ==============================================================================
r2 = r2_score(y_true, y_pred)
mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)

mask_non_zero = y_true != 0
if np.sum(mask_non_zero) > 0:
    mape = mean_absolute_percentage_error(y_true[mask_non_zero], y_pred[mask_non_zero])
else:
    mape = float("nan")

# ==============================================================================
# AFFICHAGE DU RAPPORT
# ==============================================================================
print("\n" + "=" * 60)
print(" 📊 NOUVELLES MÉTRIQUES EN DIRECT (DEPUIS VOS DONNÉES BRUTES) ")
print("=" * 60)
print(f"Lignes évaluées          : {len(df_eval):,}".replace(",", " "))
print(f"Colonne Réel/Historique  : {col_historique}")
print(f"Colonne Prédiction (IA)  : {col_realiste}")
print("-" * 60)
print(f"📈  R² (Score global)    : {r2:.4f}  ({r2 * 100:.2f}%)")
print(f"🎯  MAPE (Erreur relative): {mape * 100:.2f}%" if not np.isnan(mape) else "🎯  MAPE                 : Indisponible (division par 0)")
print(f"📉  MAE (Erreur moyenne) : {mae:,.2f} pièces".replace(",", " "))
print(f"🔢  MSE (Variance)       : {mse:,.2f}".replace(",", " "))
print(f"📐  RMSE (Écart-type)    : {rmse:,.2f} pièces".replace(",", " "))
print("=" * 60)
```
